```python

grouped_data = data.groupby(['striker', 'match_id'])['runs_off_bat'].sum()

# Create a new DataFrame from the grouped data
new_df = grouped_data.reset_index()

# Print the new DataFr
new_df.tail(20)
```

```python


# Get the list of all players
all_players = data['striker'].unique()

# Calculate the occurrences of fifties for each player
fifties_occurrences = new_df[(new_df['runs_off_bat'] >= 50) & (new_df['runs_off_bat'] < 100)].groupby('striker').size().reset_index(name='50s')

# Calculate the occurrences of centuries for each player
centuries_occurrences = new_df[new_df['runs_off_bat'] >= 100].groupby('striker').size().reset_index(name='100s')

# Merge the occurrences of fifties and centuries for all players
result = pd.DataFrame({'striker': all_players})
result = result.merge(fifties_occurrences, on='striker', how='left').fillna(0)
result = result.merge(centuries_occurrences, on='striker', how='left').fillna(0)

result
```

```python
merge_ball_df = pd.merge(merge_ball_df,result,on='striker',how='inner')
merge_ball_df
```

```python
HS = new_df.groupby('striker')['runs_off_bat'].max().reset_index()
```

```python
merge_ball_df = pd.merge(merge_ball_df,HS, on='striker',how='inner')
```

```python
merge_ball_df.rename(columns={'runs_off_bat_y':'HS'} , inplace =True)
merge_ball_df
```

```python


# Assuming you have a DataFrame called 'ball_by_ball' with columns 'Player' and 'Runs_Scored'

# Example DataFrame structure:
# ball_by_ball = pd.DataFrame({'Player': ['Player1', 'Player2', 'Player1', 'Player2', ...],
#                              'Runs_Scored': [1, 0, 4, 6, ...]})

# Load your ball-by-ball data into a DataFrame

# Group by player and concatenate all the scores
player_scores = data.groupby('striker')['runs_off_bat'].apply(list)

# Print the scores of individual players
for player, scores in player_scores.items():
    print(f"Player: {player}, innings: {scores}")
```

```python


# Assuming you have already performed the previous steps and have the 'player_scores' DataFrame

# Define dictionaries to store the count of fours and sixes for each player
fours_count = {}
sixes_count = {}

# Iterate through each player's scores
for player, scores in player_scores.items():
    # Count the number of fours and sixes for the current player
    fours_count[player] = scores.count(4)
    sixes_count[player] = scores.count(6)

# Print the counts of fours and sixes scored by each player
print("Number of Fours Scored by Each Player:")
for player, count in fours_count.items():
    print(f"Player: {player}, Fours: {count}")

print("\nNumber of Sixes Scored by Each Player:")
for player, count in sixes_count.items():
    print(f"Player: {player}, Sixes: {count}")
```

```python


# Assuming you have already performed the previous steps and have the 'fours_count' and 'sixes_count' dictionaries

# Convert the dictionaries to DataFrames
fours_df = pd.DataFrame(list(fours_count.items()), columns=['striker', '4s'])
sixes_df = pd.DataFrame(list(sixes_count.items()), columns=['striker', '6s'])
fours_df
```

```python
merge_ball_df = pd.merge(merge_ball_df,fours_df, on='striker' , how='inner')
```

```python
merge_ball_df = pd.merge(merge_ball_df,sixes_df, on='striker',how='inner')
```
