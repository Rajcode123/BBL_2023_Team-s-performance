```python
all_out_count = data.groupby(['batting_team','match_id'])['wicket_type'].count().reset_index()
```

```python
all_out_count.head(50)
```

```python
import pandas as pd

# Example DataFrame containing all team names
all_teams = data['batting_team'].unique()
result = pd.DataFrame({'batting_team':all_teams})

# Perform a left join with all_teams to include all teams
all =  all_out_count[all_out_count['wicket_type'] == 10].groupby('batting_team').size().reset_index()
all.columns = ['batting_team','10W']
result = result.merge(all,on='batting_team',how='left').fillna(0)

# Replace missing values in '10W' column with 0
result
```

```python
batting_team_df = pd.merge(batting_team_df,result,on='batting_team',how='inner')
```

```python
batting_team_df
```

```python
batting_team_df['AVG'] = (batting_team_df['runs_off_bat']/batting_team_df['wicket_type']).round(2)
```

```python
batting_team_df
```

```python
HS = data.groupby(['batting_team','match_id'])['runs_off_bat','extras'].sum().reset_index()
```

```python
HS.head(20)
```

```python
max_runs_idx = HS.groupby('batting_team')['runs_off_bat'].idxmax()

# Use the index to retrieve the corresponding values of extras
HS_max = HS.loc[max_runs_idx, ['batting_team', 'runs_off_bat', 'extras']]

HS_max
```

```python
batting_team_df = pd.merge(batting_team_df,HS_max,on='batting_team',how='inner')
```

```python
batting_team_df['HS'] = batting_team_df['runs_off_bat_y'] + batting_team_df['extras']
```

```python
batting_team_df
```

```python
batting_team_df.drop(columns=['runs_off_bat_y','extras'],inplace=True)
```

```python
batting_team_df.rename(columns={'batting_team':'TEAMS','runs_off_bat_x':'R','ball':'B','dots':'DOTS','innings_played':'I','tot_boundary':'BOUNDARY','wicket_type':'OUTS'},inplace=True)
```

```python
batting_team_df
```

```python
extra_bowled = data.groupby(['batting_team'])['wides','noballs'].sum().reset_index()
```

```python
extra_bowled
```

```python
batting_team_df.rename(columns={'TEAMS':'batting_team'},inplace=True)
```

```python
batting_team_df =batting_team_df.merge(extra_bowled,on='batting_team',how='inner')
```

```python
batting_team_df
```

```python
batting_team_df['CB'] = batting_team_df['B'] - (batting_team_df['wides'] + batting_team_df['noballs'])
```

```python
batting_team_df
```

```python
batting_team_df.drop(columns=['B'],inplace=True)
```

```python
batting_team_df.rename(columns={'CB':'B'},inplace=True)
```

```python
batting_team_df['4Runs'] = (batting_team_df['4s']*4)
```

```python
batting_team_df['6Runs'] = (batting_team_df['6s']*6)
```

```python
batting_team_df
```

```python
batting_team_df['Boundary %'] = (((batting_team_df['4Runs'] + batting_team_df['6Runs'])/batting_team_df['R'])*100).round(2)
```

```python
batting_team_df['Dot%'] = ((batting_team_df['DOTS']/batting_team_df['B'])*100).round(2)
```

```python
batting_team_df['BpB'] = (batting_team_df['B']/(batting_team_df['4s']+batting_team_df['6s'])).round(2)
```

```python
batting_team_df
```

```python
batting_team_df['BpD'] = (batting_team_df['B']/batting_team_df['OUTS']).round(2)
```

```python
batting_team_df['strike Rotation%'] = (((batting_team_df['R'] - (batting_team_df['4Runs'] + batting_team_df['6Runs'])) / batting_team_df['R'])*100).round(2)
```

```python
batting_team_df['Rotation SR'] = (((batting_team_df['R'] - (batting_team_df['4Runs'] + batting_team_df['6Runs'])) / (batting_team_df['B'] - batting_team_df['DOTS'] - batting_team_df['BOUNDARY']))*100).round(2)
```

```python
batting_team_df
```

```python
batting_team_df.drop(columns=['BOUNDARY','wides','noballs','4Runs','6Runs','10W'],inplace=True)
```

```python
batting_team_df = batting_team_df[['batting_team','I','R','B','SR','AVG','DOTS','OUTS','HS','4s','6s','Boundary %','Dot%','BpB','BpD','strike Rotation%','Rotation SR']]
```

```python
batting_team_df
```

```python
batting_team_df.to_excel('Master data index.xlsx',index=False)
```
